```python
import random
from datetime import date, timedelta

MAGASINS = ["Port-Louis", "Paris", "Lyon", "Lille"]
PRODUITS = {"Café": 2.5, "Thé": 3.0, "Sandwich": 5.5, "Salade": 7.0, "Jus": 3.5}

def generer_fichier(numero, nb_lignes=200):
    jour = date(2026, 10, 1) + timedelta(days=numero - 1)
    lignes = ["id_vente,date,magasin,produit,quantite,prix_unitaire"]
    for i in range(nb_lignes):
        produit = random.choice(list(PRODUITS))
        lignes.append(
            f"{numero}-{i},{jour},{random.choice(MAGASINS)},{produit},"
            f"{random.randint(1, 5)},{PRODUITS[produit]}"
        )
    chemin = f"/Volumes/workspace/ventes/depot/ventes_{numero:03d}.csv"
    with open(chemin, "w") as f:
        f.write("\n".join(lignes))
    print("Fichier créé :", chemin)

for n in (1, 2, 3):
    generer_fichier(n)
```
