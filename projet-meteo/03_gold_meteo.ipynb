```python
spark.sql("USE workspace.meteo")
display(spark.sql("""
    SELECT ville, DATE(heure) AS jour, temperature, precipitation
    FROM silver_weather
    ORDER BY ville, heure
    LIMIT 10
"""))
```

```python
display(spark.sql("""
SELECT
    ville,
    DATE(heure) AS jour,
    ROUND(MIN(temperature), 1) AS temp_min,
    ROUND(MAX(temperature), 1) AS temp_max,
    ROUND(AVG(temperature), 1) AS temp_moy,
    ROUND(SUM(precipitation), 1) AS pluie_totale,
    COUNT(*) AS nb_mesures
FROM silver_weather
GROUP BY ville, DATE(heure)
ORDER BY ville, jour
"""))
```

```python
spark.sql("""
    CREATE OR REPLACE TABLE gold_daily_weather AS
    SELECT
        ville,
        DATE(heure)                    AS jour,
        ROUND(MIN(temperature), 1)     AS temp_min,
        ROUND(MAX(temperature), 1)     AS temp_max,
        ROUND(AVG(temperature), 1)     AS temp_moy,
        ROUND(SUM(precipitation), 1)   AS pluie_totale,
        COUNT(*)                       AS nb_mesures
    FROM silver_weather
    GROUP BY ville, DATE(heure)
""")

print(spark.table("gold_daily_weather").count(), "lignes")
print(spark.table("gold_daily_weather").columns)
```
